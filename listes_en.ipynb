# Subscripts or how to access individual elements of a list

Each element of a list is associated with an integer called index which indicates at what position the element is in the list starting counting at 0. Thus, the element “legionary” carries the index 0 of the list called camp and “gaul” the index 1. We can access an element by doing for example: the name of the list followed by the index in square brackets

```python
print(camp[1])
```

or things like:


```python
print("There are",numbers[0]," ",camp[0]," against ",numbers[1]," ",camp[1])
```

We see how **through the use of the concept of index, we can establish correspondences between 2 lists via indices**….. And you will see that this is very, very used in science (and not only that)


We can also use indices not only to access an element but also to modify it. For example, let's imagine that the number of legionnaires is not 300 but 350. Just do `effective[0]=350` (because the index 0 corresponds to the number of legionnaires). No need to rewrite the entire list to modify a single element.

```python
numbers[0]=350
print(numbers)
```

## Negative indices

And yes, items in a list can carry negative indices. Weird right? In fact, they correspond to counting backwards. The last element in the list has the index -1, the penultimate -2 and so on. This is not necessarily useful very often except for going through a list in reverse, but **it is especially useful for very quickly accessing the last element which therefore bears the index -1**, whatever the list length.

```python
print(camp[-1])  # the index -1 allows access to the last element of a list
print(camp[-2])
```

# Slices or how to access a part of a list

Here we are discussing a very important subject that you will really need to master. It is a little complicated at first but when you practice it, you realize that it is not at all.

First of all, remember that in a list, each element has a number that corresponds to its position in the list. The count starts at 0 (see the section on [indices](https://colab.research.google.com/drive/1lUxjYly6hvAcuTTfLm77QdAVj2PwTq1U#scrollTo=c-m_lRmDBhGJ)).


Let's take our list for example: `camp=['legionary', 'gaul', 'dog', 'centurion', 'decurion']`

Extracting from this list a slice which goes **from the element of index p to that of index n** amounts to writing a **slightly special index in the form p:n+1**. I know it's strange to end with the index n+1 rather than n, but that's how it is and it even has a very practical side... You will realize it when you have written a lot of lines of code in python 😀.

Let's say we want to display the first 4 elements of the `camp` list

```python
print(camp[0:4])
```

And yes, 'legionary' is indeed the element with index 0 and 'centurion' is the element with index 3 (and not 4!).

Now let's create a new list `subCamp` in which we want elements 1 to 3 of `camp`.

```python
subCamp=camp[1:4]
print(subCamp)
```

We can also omit one of the 2 limits. In this case, if we want all the elements after that of index 3 inclusive

```python
print(camp[3:])
```

or all elements before

```python
print(camp[:3])
```

we can even take everything while omitting the 2 limits:

```python
print(camp[:])
```

You can even choose a step between the elements by adding the step in the index using the `n:p:step` syntax. For example, if we want all the elements whose indices are even up to and including index 4:

```python
print(camp[0:5:2])
```

So, a very practical shortcut for reversing a list is the ::-1 syntax in which we do not specify p, so p=0 by default, we do not specify n, so n will be the end of the array by default, and we give a step of -1 to read from right to left:

```python
print(camp)
print(camp[::-1])
```

## generate lists of integers automatically: range() with list()

In this section we will generate lists of integers automatically. To do this, there is a special type of python called `range` which contains series of integers. To create a series from 0 to 9:

```python
range(10)
```

We see that the return of the order is a little special. We don't see the list of integers but just `range(0, 10)`. In fact, this comes directly from the fact that range is a type of python, moreover we can see it clearly by doing:

```python
a=range(10)
type(a)
```

But then how do we know the values ​​contained in the list:

```python
a=list(range(10))
print(a)
```

Here we used the list function which converts certain types of Python data into a list... Which is very appropriate in the present context 😀

When using range in the form `range(n)`, the range contains all integers from 0 to n-1.

In the form `range(p,n)`, the range contains all the integers from p to n-1.

```python
a=list(range(2,10))
print(a)
```

In the form `range(p,n,inc)`, the range contains all the integers from p to n-1 with a step of inc between each value

```python
a=list(range(2,10,3))
print(a)
```

You will see in other parts of this course that range has many uses other than just generating lists.

We can also make lists “in reverse”

```python
a=list(range(10,-5,-3))
print(a)
```
