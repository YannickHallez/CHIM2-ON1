### Should you use a spreadsheet or Python? 

A spreadsheet (Microsoft Excel, LibreOffice Calc Spreadsheet, etc.)  
* is easy to install, 
* easy to use, 
* very widespread, 
* allows you to do basic calculations and simple graphs even without training,
* it allows you to directly see the values, but not the formulas, 
* is slow and does not allow working with large amounts of data,
* makes debugging complicated.

Python (or Matlab, Scilab...) is a programming language, so 
* it may seem more complicated at first, 
* it must be installed and configured before starting
* it hides the data in variables, and you therefore need to know (a little) how to program,
* it allows you to work quickly with large volumes of data,
* it allows you to directly see the formulas, but not the values, so it is more readable and easier to debug than spreadsheet formulas,
* it allows you to make better quality graphics,
* it has specialized scientific modules to help you with technical tasks. 

In summary, spreadsheets are good for doing small calculations and quick graphs even without really having any training, but they are quite limited to that. They directly show the values, but not the formulas. They are well suited for quotes, pay slips, etc.

Python is much faster and feature-rich, especially for science, but you have to take the time to learn to be comfortable with it. It allows formulas to be read quickly, which helps find errors quickly. It is therefore better suited for science than spreadsheets. 

Because of new technologies around data (Artificial Intelligence, cloud computing, etc.), more and more employees are starting to know how to program. Learning to master a programming language is therefore also an essential career development to integrate into a scientific curriculum.


### Python or another programming language?

There are several programming languages ​​used in science to process data or create models, and with capabilities similar to Python. Among the best known: Matlab, Octave, Scilab... We will talk about languages ​​like C or Fortran a little later. 

Matlab was one of the first languages ​​for science and is widely used in industry. The major downside is that it costs money and is expensive. Octave is an opensource clone of Matlab. It works well but is slow, and is rarely used. Scilab is also opensource, strongly resembles Matlab, but without being 100% compatible, and does not have as many features. Python is opensource and has an impressive number of specialized scientific modules that will chew up your work (more than Matlab). It is also a general language before being used for science. We can therefore use it to carry out many different tasks, beyond what Matlab can do. Finally, it is very easy to learn and has a very large community of users, which often allows you to find a solution to your problem on the internet. For all these reasons, we chose to use Python for this Digital Tools for Chemistry course. 

What about other well-known programming languages ​​like C, Java, or Fortran? To understand, we must introduce the notion of compiled or interpreted language. To be executed by the processor, a human-readable program (in any language) must first be converted to binary. If this translation (and optimization) step is done for the entire program once and for all, before execution, we speak of a compiled language. This is the case for C, C++, Fortran... Sometimes, your language is associated with an interpreter which reads the lines of the program during execution, and converts them and sends them to the processor on the fly. This is called interpreted language. Examples are Python, Matlab, PHP, Perl... 

The advantage of compiled languages is that execution is much faster since the only work that remains to be done during execution is the calculation that has been programmed. To make an equivalence, imagine that you (the processor) receive a paper with a choreography to perform (the program), but it is written in Chinese (the language). You have two solutions: either you translate it beforehand and write it on another paper (compiled language), or you ask a friend to translate it for you on the fly at the same time as you dance (interpreted language). We can imagine that the first version will allow you to dance much faster, right? But then why use interpreted language? Essentially because they are much easier to use. For a compiled language program to work well, many things must be specified in advance, such as how memory is used, for example. With an interpreted language, this sort of thing is adapted automatically at runtime, but at the cost of reduced performance. 

There is no absolute best option between compiled or interpreted languages. They are different tools, useful for different tasks. For applications that are very resource intensive, for which performance is essential, compiled languages ​​are used (simulation software, video games, operating systems). For less performance-intensive applications, the simplicity of interpreted languages ​​is interesting. 

In practice, it is possible to mix these languages ​​and we are seeing more and more hybrid applications with the user interface supported by an interpreted language and the heavy computing parts carried out using a compiled language.
