# What is an array?

Arrays are very similar to lists... and even at first you'll probably confuse them 😞. However, they do not have the same usefulness. While lists allow you to store data of all types, tables are intended to carry out calculations in the mathematical sense of the term, and in particular on vectors or matrices. 

`(numpy) array`, like `list`, `float`, `range`, `int` ... is a type of variables.

# How to define an array?

1. **Import the numpy module**, what you do implicitly when you import `pylab`.

2. **Define an array**. For this, it works a bit like lists with values ​​in square brackets but you include this list in a call to `array()`.


```python
from pylab import *

a = array([1,2.5,-3.9])
print(a)
print(type(a))
```

Besides, we can very well convert a list into an array

```python
l = [1,2.5,-3.9]    # here is a list
b = array(l)    # we convert the list into an array
print(b)
print(type(b))
```

In fact, the essential difference between a list and the corresponding array is that 
* a **one-dimensional** array can essentially be considered a **vector** in mathematics 
* a **2-dimensional array** like a **matrix** in mathematics (we won't talk about it for the moment)

So, as these objects have mathematical meaning, we can do mathematical operations with them

```python
# examples of operations
c=b+1 # add a scalar, like for mathematical vectors or matrices
print(b)
print('addition of a scalar',c) 
print('addition of two arrays:',b+c) # addition of 2 arrays
# we could do the same with subtraction
d=b*2 # multiply by a scalar
print('multiplication by a scalar:',d)
print('multiplication of 2 arrays:',b*c)
# we could do the same with the division

# We did calculations with two-dimensional matrices, but it also works with vectors (1-dimensional array)
```

# **VERY IMPORTANT NOTES\:**

The operations between 2 matrices that we have just performed are **member to member operations**, that is to say that an element of the first vector at a given position is added or multiplied by the element at the same position in the second vector. Here, nothing surprising for addition, but on the other hand, it is very important to understand for multiplication and division. Indeed, **in math, we cannot multiply two vectors of the same dimension** and vector division does not exist as such.

# Some ways to generate arrays automatically (arange, linspace, zeros, ones)

## arange

For arrays, there is a parallel to the `range` function that we saw for lists: it is the `arange` function.

So, if we want to automatically generate a list of integers from 10 to 20:


```python
print('with a step of 1:',arange(10,20)) 
print('with a step of 2:',arange(10,20,2)) 
```

We find the fact that the values ​​contained in the array stop at the value preceding the value 20 and not at 20.

Unlike range, we can also work with real numbers

```python
r=arange(0,1,5e-2)
print(r)
```

## linspace

Another way to generate real values equally distributed over an interval is to use the `linspace` function. This function takes 3 parameters, the first 2 being the start point and the end point. The last is the number of points we want on the interval between the start and finish points.

```python
e=linspace(0,1,11)
print(e)
```

We can clearly see the differences with arange. Here, the end point is indeed part of the vector unlike arange. Furthermore, while arange needs the distance between each value, linspace needs the number of values ​​including extreme values. it is also because these 2 extreme values ​​are included that I took 11 points to have 0.1 between each value.

## arange or linspace? That is the question...

Indeed, you have seen that these two functions do almost the same thing: discretize an interval. So which one to choose?

It actually basically depends on what you want to use as your third parameter. If you know the number of points to put on the interval, go for `linspace`. If you know the desired distance between each value, you must choose `arange`.

For arange, there is even a way to ensure that if the last value is part of the set of values ​​to reach.

```python
r=arange(0,1,5e-2)
print(r)  # 1 is not part of the vector with arange
r=arange(0,1+5e-2,5e-2)
print(r) # 1 is now part of it... we kind of fooled arange
```

# Array copy

Here we come across a subject which may not be easy to understand for a beginner and which sometimes leads to errors which are very difficult to identify.

Consider the simple example below where I put a vector into a variable a. Then, I decide to also put it in another variable b like that.

```python
a=array([12,15.5,-2])
b=a
```

Legitimately and as in many programming languages, we could think that b contains the same values ​​as a but is completely independent of it. Let's check modifying a value of b like for example,

```python
b[1]=203
print(b) # there it is no surprise
print(a) # this is much more surprising...
```

And yes, we modified b, but the vector a was also modified when we modified b. The simplest way to make b completely independent of a while copying its values ​​is to generate a new list from a.

```python
a=array([12,15.5,-2])
b=array(a)
b[1]=203
print(b) # there it is no surprise
print(a) # we didn't touch that
```

It should be noted that lists behave in the same way with respect to copying;

# Some useful functions (max, min, argmax, argmin, sum, mean)

The `min` and `max` functions allow you to respectively know the minimum value and the maximum value in an array while `argmin` and `argmax` allow you to access the index of the minimum value and that of the maximum value respectively. 

`sum` returns the sum of the elements.
`mean` returns the average value of the elements

```python
a=array([10,1,15,12.5,18,19.5,5,12])
print(min(a))
print(argmin(a))
print(max(a))
print(argmax(a))
print(sum(a))
print(mean(a))
```

# Boolean array masking

In python we can easily extract or modify data in an array using relational operators

```python
b = a>=10
print(b)
print(a[b])
```

`b` contains as many booleans as `a`, and it checks if each element of `a` is greater than or equal to 10. 

The call to a[b] gives a new array with obly the values greater than 10. 

You can also modify data using this type of operation.

```python
a[a<10]=0
a[a>=10]=1
print(a)
```

# access the dimensions of 2-dimensional arrays:

We find the same commands as for lists, namely `ndim`, `shape` and `size`. To fully understand what is behind arrays, especially in 2 dimensions, we will detail a few points. Let's start by observing the result of these functions with a vector.

```python
a = array([1,2.5,-3.9])
print(a)
print('dimension:',ndim(a))
print('shape :',shape(a))
```

`ndim` tells us the number of dimensions of a. It is therefore a vector. `shape` gives us the number of elements in a slightly strange form. This is another Python object called `tuples` which corresponds to a series of data, surrounded by parentheses, and separated by commas. We will not go into more detail in the description of this type of object. But, in the present context, we just need to know that `shape` gives the number of elements for each dimension. And in this case, as `a` has only one dimension, `shape` only returns a number which is the number of elements of the single dimension.

Let us now look at the structure of a 2-dimensional matrix. I'll take this opportunity to show you another way to call the `ndim` and `shape` functions.

```python
c = array([[1,2,3],[4,5,6]])    
print(c)
print('dimension:',c.ndim)
print('shape :',c.shape)
```

We see here that the array c has 2 dimensions and that the first of these dimensions is 2 and the second 3. Why not the opposite... well, it's a bit like nesting dolls. Let's write differently, the same matrix.

```python
a = array([1,2,3])
b = array([4,5,6])    
c=array([a,b])
print(c)
```

We see that in fact `c` is a vector with 2 elements, each of the elements being a vector with 3 elements. The dimension that we see first is therefore 2 then comes after the next dimension: 3. A bit like when for nesting dolls where we first open the largest which is the only one we see to see the next one.

# Slices or how to access rows or columns in a 2D array

Here again, the parallel with what we saw for lists is maintained. However for lists, we have covered this part only for single-dimensional lists. We will now see how to do this for two-dimensional arrays. The same principles can be applied for lists of lists.

### Access a line

Suppose we want to access the second line of array `c`. This amounts to accessing all columns in the second line of `c`. As we saw for lists, all elements correspond to the symbol:

```python
c = array([[1,2,3],[4,5,6]]) 
print(c[1,:])
```

### Go to a column

Likewise the same way if we want to access, let's put it in the second column, this amounts to taking all the lines of the second column. We therefore use the same formalism but this time, it is the first index which takes the symbol \:

```python
print(c[:,1])
```

we notice that python does not retain the column shape! For Python, this column has only one dimension and the display is done as for any single-dimensional element.
