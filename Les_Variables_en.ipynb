### What is a variable

A variable is used to store information in computer memory. It has a name, for example `x`, which hides a value, for example `3`, and an address, that is to say the place where it is located in memory. 

To create a variable in Python, simply give it a value. For example :

```python
x = 3
```

Python will automatically identify that `3` is an integer and therefore reserve space in memory to store an integer. It will set the **type** of variable to **integer**. Finally, without actually knowing the address of `x` in memory, we will be able to access its value using the variable name `x`. For example to display the stored value:

```python
print(x)
```

NB: we could also have simply typed

```python
x
```

Be careful, using `=` in python is not the same as in mathematics. The `=` symbol does not establish a logical link. It allows you to first calculate the value to the right of the `=` and store it in memory at the address of the variable given on the left. We're talking about assignment. The value `3` has been assigned to the variable `x`. So, in the following example...

```python
x = 2
y = x 
print("First display of y:", y) 
x = 3 
print("Second display of y:", y)
```

We create `x` in the first line with the value `2`. In the second line, we create a variable `y` and store the value to the right of the `=`, that is to say `2`. In the third line we check that this is indeed the case. In the fourth line, we place the value `3` in the variable `x`, then in the fifth line, we display the value of `y`. We see that it is still at `2`, while the value of `x` has changed to `3`. This is because the second line `y=x` is not a logical connection between `x` and `y`. It simply means “take the value at the address of `x`, and copy that value to the address of `y`”. If we change the value of `x` afterwards, `y` will not change. 

There are many types of variables in Python, as in all languages. You will use **integers** (`int`), **reals** (`float`), **strings** (words or sentences, `str`), and **booleans** (true or false, `bool`). 

You can check the type of a variable or an expression using the `type()` function.

```python
print( type(3) )
print( type(6.5) )
print( type("Hello!") )
print( type(False) )
print( type(True) ) 
```

The first two examples are intuitive. For real numbers, we use the point as a decimal separator. In the third, we see that a character string is defined by putting quotes around something. In our example we see that `“Hello!”` is a character string composed of the characters H, e, l, l, o and !. `False` and `True` are keywords which represent true or false booleans in python. Their type is therefore `bool`.

Let's define a variable and check its type:

```python
a = 5.9
print( type(a) )
print( type("a") )
```

```python
b = "coucou"
print( type(b) )
```

When we create a variable, we can use one or more characters including lowercase letters, uppercase letters, numbers, but no spaces or special characters except `_` (underscore) which is very practical precisely to give the illusion of a space. You don't have to start with a number. Possible variable names are: `x`, `xyz`, `house`, `X`, `my_variable`, `the_price_of_bread`. NOT possible variable names are: `my dog's name` (no spaces), `42street` (starts with a number), `a$bc` (special character), `µ` (special character). 

Python is case sensitive, that is:
* `x` and `X` are recognized as two different variables. 
* Language syntax elements must be case sensitive. Booleans `True` and `False` are always capitalized, the function `print()` must always be in lowercase, the function `Circle()` must always be capitalized, etc. 

Beyond these strict rules on variable names, you must always be careful to give explicit names that another person can easily understand. For example, to store the value of the velocity of a gas in a chromatography column, we will avoid `x34f6mv`, or even `v`, and we will prefer something like `gasVelocity`. 

### Variable operations 

It is possible to perform operations on numeric types (`float`, `int`), but also on character strings (join 2 words together, etc.), or on Booleans. 

Here are the most important operations concerning numbers: 
* Usual operations: + - * /            
* Power: ** 
* Quotient of a Euclidean division: // 
* Remainder of a Euclidean division (modulo): %

```python
print( "3+5 =", 3+5 )
print( "9*3.0 =", 9*3.0 )
print( "5/2 =", 5/2 )
print( "4**2 =", 4**2 )
print( "2**(-1) =", 2**(-1) )
print( "10//3 =", 10//3 )
print( "10%3 =", 10%3 )
```

Here are the most common operations concerning character strings:
* Two strings can be concatenated with +
* A string `X` multiplied by an integer `N` creates a new string in which the string `X`is copy/pasted `N` times.

```python
print( "cou"*2 )
print( "Tra" + 2*"la" + "!" )
```

For real numbers, we can use scientific notation, for example 3.23 10<sup>5</sup> is written `3.23e5`, 1.23 10<sup>-4</sup> is written `1.23e-4`. You should NEVER use `1.23*10**(-4)` because this last example, although generating the correct value, requires unnecessary work from the processor (a multiplication and a power) and is a source of  errors. 

The priorities of the operations are the same as in mathematics: * and / have priority over + and -. ** takes priority over the rest. You can use as many parentheses as necessary. 

You can perform operations between different numeric types, for example an integer and a real. Python will give a result with the “richest” type. For example `5+2` will give `7` (int+int gives int), and `5+2.0` will give `7.0` (int + float gives float). 

We sometimes want to change the type of a variable. `int(X)` converts `X` to int, `float(X)` converts `X` to real, `str(X)` converts `X` to string. See the example below:

```python
a = 3.6 
print( type(a) )

b = int(a) 
print( type(b) )
print( b )

c = str(a) 
print( c )

print( a+b )

print( a+c )
```

We see that `int(3.6)` is not just a conversion. Indeed, as the result must be integer, it will be the integer part of the number passed as an argument.

We then see that `a+b` was an operation between `int` and `float`, with a result `float`. But `a+c` is an operation between a `float` (a) and a string `c`, which makes no sense and Python reminds us with an error message: “the + operator cannot accept a float and a str”. 

It is often necessary to convert numeric values ​​such as real numbers into character strings, for example for graph legends. The str() function does not control the format of the created string. Here is an example in which we see that str() will keep the 16 digits after the decimal point:

```python
from numpy import pi
print( str(pi) )
```

To take control of the format of a character string representing a number, we can use f-strings. Here is an example which will be explained next:

```python
print( f"The value of pi is close to {pi:8.5f}") 
```

In an f-string you can write text as in a normal character string. For example f"hello" is the same as "hello". But we can also incorporate numerical values ​​from program variables using the {variable:format} syntax. In the example the variable is pi, of type float, and the format is 8.5f. To know all the details of the formats you have to read the documentation, but in short, the first digit is the maximum number of characters used to represent the value (8 here), the second digit is the number of digits after the decimal point (5 here), and the letter at the end says whether you want a float (f) or engineer (e) or integer (d) display. There are other subtleties but with this you have the most useful part of f-strings. Here is another example in engineer format:

```python
print( f"The value of pi is close to {pi:3.2e}") 
```

### Reading variables from the keyboard 

If we want to create a variable whose value is typed on the keyboard at run time, we can use the `input()` function. Please note, this function always defines a character string variable. Try the following example by typing a word, and typing a number. You will always have the type <class 'str'>.

```python
x = input()
print( "You entered:", x, " and variable x has type ", type(x))
```

If we want to do a calculation, that will pose problems. For example :

```python
a = input()
b = input()
a+b
```

Since a is a word, for example '2', and b is a word, for example '4', the + symbol will paste (concatenate) these two words to form a new word '24'. To perform calculations, you must convert the variables "a" and "b" as we saw above.

```python
a = input()
a = int(a)
b = input()
b = int(b) 
a+b
```

For greater readability, we generally directly chain the reading with `input` and the conversion on the same line. For example:

```python
a = int( input() ) 
b = float( input() )
a+b
```

When 2 functions are called one inside the other (here `input()` inside `int()` or `float()`, Python always executes the inner function first, and then the outer function. So here, in the first line, the `input()` is executed and you enter a value which is a 'str' in memory. This 'str' is passed directly to `int()` which converts it to an integer. Same thing in the second line but the conversion is towards a real.    

We often want to display a message on the screen to tell the user what to do. We can do this by putting the message as an argument to the `input()` function.

```python
x = input("Please enter a value: ")
print( x )
```
